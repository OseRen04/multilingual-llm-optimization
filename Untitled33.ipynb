```python

# ============================================================
# Qwen3-8B：10言語のトークン使用頻度分析
# ============================================================
#
# 【目的】
# 日本語・英語・ロシア語・韓国語・中国語の5言語を保持し、
# その他の言語に使われやすい語彙の削減可能性を調査する。
#
# 【分析対象】
# 保持対象：日本語、英語、ロシア語、韓国語、中国語
# 削減候補：アラビア語、フランス語、スペイン語、
#           ヒンディー語、タイ語
#
# 【分析条件】
# 1言語あたり500万トークン。
# 10言語で最大5,000万トークンを集計する。
#
# 【出力】
# 1. 各言語のトークン使用頻度
# 2. 言語別の語彙使用数
# 3. 保持対象と削減候補の共有トークン数
# 4. 削減候補のトークンID一覧
#
# 【注意】
# このコードは語彙使用頻度の分析まで実行する。
# モデルの重みを削除・変更する処理ではない。
# ============================================================


# ------------------------------------------------------------
# 1. ライブラリ
# ------------------------------------------------------------

# トークンの出現回数を数える。
from collections import Counter

# 保存先ディレクトリを操作する。
from pathlib import Path

# JSONファイルを保存する。
import json

# CSVファイルを保存する。
import csv

# Hugging Faceのデータセットを読み込む。
from datasets import load_dataset

# Qwen3-8BのTokenizerを読み込む。
from transformers import AutoTokenizer

# Google DriveをColabに接続する。
from google.colab import drive


# ------------------------------------------------------------
# 2. Google Driveを接続
# ------------------------------------------------------------

# Colabの一時ストレージではなく、
# 自分のGoogle Driveに分析結果を保存する。
#
# 初回はGoogleアカウントの認証が求められる。
drive.mount("/content/drive")

# 今回の分析結果の保存先。
OUTPUT_DIR = Path(
    "/content/drive/MyDrive/"
    "multilingual-llm-optimization/"
    "vocabulary_analysis_10langs"
)

# フォルダが存在しない場合は作成する。
OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


# ------------------------------------------------------------
# 3. 実験設定
# ------------------------------------------------------------

# Qwen3-8BのTokenizerを使用する。
# 80億パラメータのモデル本体は読み込まない。
MODEL_ID = "Qwen/Qwen3-8B"

# 各言語で分析するトークン数。
TARGET_TOKENS_PER_LANGUAGE = 5_000_000

# データの順序を混ぜる際の乱数シード。
SEED = 42

# ストリーミングデータのシャッフル用バッファ。
# 完全な無作為抽出ではない。
SHUFFLE_BUFFER_SIZE = 1000

# 1文書あたり分析する最大文字数。
# 極端に長い文書による処理時間の偏りを抑える。
MAX_CHARACTERS_PER_DOCUMENT = 10000


# ------------------------------------------------------------
# 4. 分析対象のデータセット
# ------------------------------------------------------------

# 保持対象5言語。
#
# 中国語のcmn_Haniは簡体字・繁体字が混在し得る。
# 簡体字限定の分析は別途フィルタリングが必要。
KEEP_LANGUAGES = {
    "日本語": ("HuggingFaceFW/fineweb-2", "jpn_Jpan"),
    "英語": ("HuggingFaceFW/fineweb", "sample-10BT"),
    "ロシア語": ("HuggingFaceFW/fineweb-2", "rus_Cyrl"),
    "韓国語": ("HuggingFaceFW/fineweb-2", "kor_Hang"),
    "中国語": ("HuggingFaceFW/fineweb-2", "cmn_Hani"),
}

# 削減候補5言語。
REMOVE_LANGUAGES = {
    "アラビア語": ("HuggingFaceFW/fineweb-2", "arb_Arab"),
    "フランス語": ("HuggingFaceFW/fineweb-2", "fra_Latn"),
    "スペイン語": ("HuggingFaceFW/fineweb-2", "spa_Latn"),
    "ヒンディー語": ("HuggingFaceFW/fineweb-2", "hin_Deva"),
    "タイ語": ("HuggingFaceFW/fineweb-2", "tha_Thai"),
}


# ------------------------------------------------------------
# 5. Tokenizerを読み込む
# ------------------------------------------------------------

# Hugging FaceからTokenizerを取得する。
tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)

# Tokenizerが扱うトークンIDの総数。
VOCAB_SIZE = len(tokenizer)

# 特殊トークンIDを保護対象として記録する。
# 例：会話の開始・終了・区切りなど。
PROTECTED_IDS = set(tokenizer.all_special_ids)


# ------------------------------------------------------------
# 6. 分析結果を保持する変数
# ------------------------------------------------------------

# 保持対象5言語で観測した全トークン頻度。
keep_frequency = Counter()

# 削減候補5言語で観測した全トークン頻度。
remove_frequency = Counter()

# 言語別の集計結果。
summary_rows = []


# ------------------------------------------------------------
# 7. 1言語分の文章を分析する関数
# ------------------------------------------------------------

def analyze_language(language, dataset_id, subset):

    print(f"\n分析開始：{language}", flush=True)

    # データセットをストリーミングで取得する。
    # 全文書をColabのディスクに保存するわけではない。
    dataset = load_dataset(
        dataset_id,
        name=subset,
        split="train",
        streaming=True,
    )

    # 一定件数の範囲で文書の順序をシャッフルする。
    dataset = dataset.shuffle(
        seed=SEED,
        buffer_size=SHUFFLE_BUFFER_SIZE,
    )

    # トークンIDごとの出現回数を格納する。
    frequency = Counter()

    # 分析済みトークン数。
    total_tokens = 0

    # 処理した文書数。
    total_documents = 0

    # 文書を1件ずつ取得する。
    for row in dataset:

        # 文書本文を取得する。
        text = row.get("text", "")

        # 空の文書は対象外とする。
        if not text:
            continue

        # 文書をQwen3-8BのトークンIDに変換する。
        #
        # add_special_tokens=False：
        # 会話用の特殊トークンを追加しない。
        token_ids = tokenizer.encode(
            text[:MAX_CHARACTERS_PER_DOCUMENT],
            add_special_tokens=False,
        )

        # 設定したトークン数を超えないようにする。
        remaining = (
            TARGET_TOKENS_PER_LANGUAGE - total_tokens
        )
        token_ids = token_ids[:remaining]

        # 各トークンIDの出現回数を加算する。
        frequency.update(token_ids)

        # 分析済みトークン数を更新する。
        total_tokens += len(token_ids)

        # 処理した文書数を更新する。
        total_documents += 1

        # 500万トークンに到達したら終了する。
        if total_tokens >= TARGET_TOKENS_PER_LANGUAGE:
            break

    # --------------------------------------------------------
    # 8. 言語別の頻度を保存
    # --------------------------------------------------------

    # 言語ごとに別ファイルへ保存する。
    # Colabが後で切断されても完了済み言語の結果は残る。
    result = {
        "language": language,
        "dataset_id": dataset_id,
        "subset": subset,
        "total_tokens": total_tokens,
        "total_documents": total_documents,
        "unique_tokens": len(frequency),
        "frequency": dict(frequency),
    }

    output_path = OUTPUT_DIR / f"{subset}_frequency.json"

    with output_path.open("w", encoding="utf-8") as f:
        json.dump(
            result,
            f,
            ensure_ascii=False,
        )

    print(
        f"{language}："
        f"{total_tokens:,}トークン、"
        f"{len(frequency):,}種類",
        flush=True,
    )

    # 集計結果を呼び出し元に返す。
    return frequency, {
        "language": language,
        "group": (
            "keep" if language in KEEP_LANGUAGES
            else "remove"
        ),
        "dataset": dataset_id,
        "subset": subset,
        "total_tokens": total_tokens,
        "documents": total_documents,
        "unique_tokens": len(frequency),
    }


# ------------------------------------------------------------
# 9. 保持対象5言語の分析
# ------------------------------------------------------------

for language, (dataset_id, subset) in KEEP_LANGUAGES.items():

    # 1言語分の頻度分析を実行する。
    frequency, summary = analyze_language(
        language,
        dataset_id,
        subset,
    )

    # 保持対象の全体頻度に加算する。
    keep_frequency.update(frequency)

    # 言語別の集計行を保存する。
    summary_rows.append(summary)


# ------------------------------------------------------------
# 10. 削減候補5言語の分析
# ------------------------------------------------------------

for language, (dataset_id, subset) in REMOVE_LANGUAGES.items():

    frequency, summary = analyze_language(
        language,
        dataset_id,
        subset,
    )

    # 削減候補の全体頻度に加算する。
    remove_frequency.update(frequency)

    summary_rows.append(summary)


# ------------------------------------------------------------
# 11. 保持対象と削減候補の語彙を比較
# ------------------------------------------------------------

# 保持対象5言語で使用されたトークンID。
keep_ids = set(keep_frequency.keys())

# 削減候補5言語で使用されたトークンID。
remove_ids = set(remove_frequency.keys())

# 両方のグループで使用されたトークンID。
shared_ids = keep_ids & remove_ids

# 保持対象でのみ観測されたトークンID。
keep_only_ids = keep_ids - remove_ids

# 削減候補でのみ観測されたトークンID。
remove_only_ids = remove_ids - keep_ids

# 10言語のどれでも観測されなかったトークンID。
all_ids = set(range(VOCAB_SIZE))
unobserved_ids = all_ids - keep_ids - remove_ids

# 今回の頻度情報から抽出した削減候補。
#
# 保持対象5言語では未観測で、
# 削減候補5言語では観測されたトークンを選ぶ。
#
# ただし特殊トークンは除外する。
#
# これらは「候補」であり、そのまま削除してはいけない。
candidate_ids = remove_only_ids - PROTECTED_IDS


# ------------------------------------------------------------
# 12. トークンごとの削減候補をCSV保存
# ------------------------------------------------------------

# 削減候補言語での使用頻度が高い順に並べる。
sorted_candidates = sorted(
    candidate_ids,
    key=lambda token_id: remove_frequency[token_id],
    reverse=True,
)

candidate_path = OUTPUT_DIR / "removal_candidates.csv"

with candidate_path.open(
    "w",
    encoding="utf-8-sig",
    newline="",
) as f:

    writer = csv.writer(f)

    writer.writerow([
        "token_id",
        "token",
        "keep_frequency",
        "remove_frequency",
    ])

    for token_id in sorted_candidates:

        writer.writerow([
            token_id,
            tokenizer.convert_ids_to_tokens(token_id),
            keep_frequency[token_id],
            remove_frequency[token_id],
        ])


# ------------------------------------------------------------
# 13. 言語別の集計CSVを保存
# ------------------------------------------------------------

summary_path = OUTPUT_DIR / "language_summary.csv"

with summary_path.open(
    "w",
    encoding="utf-8-sig",
    newline="",
) as f:

    writer = csv.DictWriter(
        f,
        fieldnames=[
            "language",
            "group",
            "dataset",
            "subset",
            "total_tokens",
            "documents",
            "unique_tokens",
        ],
    )

    writer.writeheader()
    writer.writerows(summary_rows)


# ------------------------------------------------------------
# 14. グループ間の重複集計を保存
# ------------------------------------------------------------

overlap_result = {
    "vocab_size": VOCAB_SIZE,
    "keep_unique_tokens": len(keep_ids),
    "remove_unique_tokens": len(remove_ids),
    "shared_tokens": len(shared_ids),
    "keep_only_tokens": len(keep_only_ids),
    "remove_only_tokens": len(remove_only_ids),
    "unobserved_tokens": len(unobserved_ids),
    "protected_special_tokens": len(PROTECTED_IDS),
    "removal_candidate_tokens": len(candidate_ids),
}

with (OUTPUT_DIR / "overlap_summary.json").open(
    "w",
    encoding="utf-8",
) as f:
    json.dump(
        overlap_result,
        f,
        ensure_ascii=False,
        indent=2,
    )


# ------------------------------------------------------------
# 15. 最終結果を表示
# ------------------------------------------------------------

print("\n========== 10言語の分析結果 ==========")

print(
    f"Qwen3 Tokenizer総語彙数："
    f"{VOCAB_SIZE:,}"
)

print(
    f"保持対象5言語で観測："
    f"{len(keep_ids):,}"
)

print(
    f"削減候補5言語で観測："
    f"{len(remove_ids):,}"
)

print(
    f"両グループで共有："
    f"{len(shared_ids):,}"
)

print(
    f"削減候補5言語でのみ観測："
    f"{len(remove_only_ids):,}"
)

print(
    f"10言語すべてで未観測："
    f"{len(unobserved_ids):,}"
)

print(
    f"特殊トークンを除外した削減候補："
    f"{len(candidate_ids):,}"
)

print("\n保存先：", OUTPUT_DIR)
```
